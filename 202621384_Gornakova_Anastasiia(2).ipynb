<font color="red">**지도를 그려서 median_house_value/population 별로 scatter하여 어느지역의 집값이 비싼지 말해줘...**</font>

```python
import pandas as pd
path = '/content/sample_data/california_housing_train.csv'
df = pd.read_csv(path)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))
plt.scatter(
x=df['longitude'],
y=df['latitude'],
s=df['population'] / 100,
c=df['median_house_value'],
cmap='viridis',
alpha=0.5
)

plt.colorbar(label='Median House Value')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('California Housing')
plt.show()
```

해안 지역의 집값이 높고, 내륙 지역의 집값이 낮다
